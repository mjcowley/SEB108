## 4.1 Booleans and comparisons

Booleans can only be in one of two states: `True` or `False`. Assign a variable to `True` or `False` below, and notice how Python syntax highlights these keywords, indicating their special role.

```python
# Assigning a boolean value to a variable
is_sunny = True

# Checking the value
print(is_sunny)
```

You'll mostly get booleans by **comparing** things. Each comparison below gives `True` or `False`.

```python
reading = 12.5

print(reading > 10)    # greater than
print(reading < 10)    # less than
print(reading >= 12.5) # greater than or equal to
print(reading == 12.5) # equal to (two equals signs)
print(reading != 0)    # not equal to
```

Watch out for `==`. One equals sign (`=`) **stores** a value in a variable. Two (`==`) **checks** whether two things are equal.

You can combine comparisons with `and` (both must be true) and `or` (at least one must be true).

```python
print(reading > 10 and reading < 20)  # is the reading between 10 and 20?
print(reading < 5 or reading > 100)   # is it unusually low or unusually high?
```

Comparisons are how you'll pick out the rows you want from a dataset in Practical 2, for example "every sample above 50 mg/L". You'll also use them to make decisions in your code (Section 5).

## 4.2 Lists

Lists are Python's most versatile containers. You can put nearly anything inside a list, including different data types or even other lists. However, the practicality of mixed-type lists can be limited : the benefit of storing a series of numbers in a list is that you can perform operations on them collectively without concern for compatibility issues. Here's how to create a list:

```python
# Defining a list
my_list = [1, "a_string", True, 3.14, [2, 4, 6]]

# Printing the list
print(my_list)
```

The list above is a bit chaotic. You'll rarely want a list with such mixed contents, but it shows that Python lets you put almost anything in one. You don't always have to type a list out by hand, either. `range()` makes a run of evenly spaced whole numbers, and `list()` turns it into a list.

```python
# Even numbers from 2 up to 20
even_numbers = list(range(2, 21, 2))
print(even_numbers)
```

`range(2, 21, 2)` means "start at 2, stop **before** 21, go up in steps of 2". The stop number is never included, which is why we wrote 21 to get 20.

Two built-in functions you'll use with lists all the time are `len()`, which counts the items, and `sum()`, which adds them up.

```python
print(len(even_numbers))
print(sum(even_numbers))
```

## 4.3 Indexing and slicing

To view the contents of a list, we simply print it:

```python
print(my_list)
```

However, there are instances where we need to extract specific elements for calculations. This is where **indexing** comes in. In Python, list indexing starts at 0, meaning the first element is at the "0th index". Here's how you can access elements using the list defined previously:

```python
my_list[0]
```

We put square brackets after the variable name, with the index we want inside. To pull out several items at once, give a range of indices instead. This is called **slicing**. The first number is included and the second isn't:

```python
my_list[0:2]
```

We can also specify a step value, which is the third number in the brackets. This sets the interval between elements in the slice (for example, 2 means take every second element).

```python
my_list[0:5:2]
```

What if we need to access an element within a nested list [2, 4, 6]? This situation calls for **double indexing**. Additionally, this is a good opportunity to introduce **negative indexing**, a handy feature that allows counting backwards from the end of a list, simplifying access to its latter elements:

```python
my_list[-1][1]
```

The above code snippet demonstrates how to access the number 4 within the nested list [2, 4, 6]. The negative index `-1` refers to the last element in the list, which is itself a list. The index `[1]` then extracts the second element from that list.

So, to summarise, **indexing** gets you one item by its position, and **slicing** gets you a run of items, like `my_list[0:2]`. Both work on lists and strings.

- **Positive Indexing**: Starts from 0 at the beginning of the iterable and increases by 1 for each subsequent element.
- **Negative Indexing**: Starts from -1 for the last element, -2 for the second last, and so on, making it easy to access elements from the end.

## 4.4 Dictionaries

Dictionaries function similarly to lists, but instead of using numeric indices to access elements, we use unique "keys" to associate with each "value". Here's an example:

```python
# Create a dictionary
my_dict = {"name": "Alice", "age": 30, "city": "Sydney"}

# Accessing a value by its key
print(my_dict["name"])
```

This Python snippet demonstrates how to create a dictionary and access its elements. Dictionaries store data in key-value pairs, allowing you to quickly access a value by specifying its corresponding key:

- **Creating a Dictionary:** We define `my_dict` with three key-value pairs, mapping "name" to "Alice", "age" to 30, and "city" to "Sydney".
- **Accessing a Value:** To retrieve a value, such as Alice's name, we use the syntax `my_dict["name"]`, specifying the key to obtain its associated value.

Dictionaries are handy when the **name** of a value matters more than its position. You look things up by key, so you never need to remember where something is stored. (Dictionaries do keep things in the order you added them, but you'll rarely need to rely on that.)

What if we wanted to add a new key-value pair to the dictionary? This is a simple task in Python:

```python
# Adding a new key-value pair
my_dict["occupation"] = "Engineer"

# Inspecting the updated dictionary
print(my_dict)
```

Great, but let's now modify the value of an existing key. To do this, we simply reassign the value to the key:

```python
# Modifying a value
my_dict["age"] = 31

# Inspecting the updated dictionary
print(my_dict)
```

As you can see, Alice is now 31 years old. This demonstrates how dictionaries are mutable, allowing you to change their contents as needed. Let's try one more change and remove a key-value pair:

```python
# Removing a key-value pair
del my_dict["city"]

# Inspecting the updated dictionary
print(my_dict)
```

Each key in a dictionary has to be unique, and keys are usually strings or numbers. You can't rename a key directly, but you can add a new one and delete the old one. The values are much more flexible: you can change, add or remove them whenever you like.

Dictionaries suit any data where each value has a meaningful label, like the details of a sampling site: its name, its location and the date it was set up.

## 4.5 Strings

We've previously encountered strings, which allow you to incorporate text (such as words or file paths) into your code that Python wouldn't inherently understand. Strings are incredibly versatile, capable of containing any character. Real data files sometimes store numbers as text, especially when a column has a stray entry like `<0.5` or `n/a` in it. When that happens you'll need to convert the text to numbers yourself, which you'll do in Practical 2. Importantly, strings are iterable, meaning they can be indexed character by character, similar to lists.

```python
# Defining a string
my_string = "Hello, world!"

# Accessing characters in the string
first_char = my_string[0]  # 'H'
last_char = my_string[-1]  # '!'

# Printing the characters
print(f"First character: {first_char}")
print(f"Last character: {last_char}")
```

This snippet illustrates several key aspects of working with strings:

- **Defining a String:** We create a simple string `my_string` with the value `"Hello, world!"` to show how text is stored.
- **Accessing Characters:** Using our knowledge of indexing, we can extract specific characters from the string. `my_string[0]` retrieves the first character (`'H'`), while `my_string[-1]` fetches the last character (`'!'`), demonstrating how strings are iterable.
- **Printing Characters:** The `print()` function displays the first and last characters. The "f" before the string literals indicates an **f-string**, which allows for the direct insertion of expressions into string literals using curly braces `{}`. This method simplifies the process of combining text and variables/data in output.

Let's see what else we can do with strings. For instance, we can concatenate strings, which means combining them into a single string. This is achieved using the `+` operator:

```python
# Concatenating strings
new_string = my_string + " from Python"
print(new_string)
```

Here, we've combined the original string `"Hello, world!"` with the phrase `" from Python"`, creating a new string that reads `"Hello, world! from Python"`. This operation is known as string concatenation. Let's now try to replace a portion of the string with another string:

```python
# Replacing a substring
replaced_string = my_string.replace("world", "Python")
print(replaced_string)
```

Here, we've replaced the substring `"world"` in the original string with `"Python"`, resulting in the new string `"Hello, Python!"`. This operation is known as string replacement. Next, let's try splitting a string into a list of substrings:

```python
# Splitting a string
split_string = my_string.split(",")
print(split_string)
```

Here, we've split the original string `"Hello, world!"` into a list of substrings, using the comma `","` as the separator. The result is a list `['Hello', ' world!']`, where the comma has been removed. This operation is known as string splitting. Finally, let's try case conversion on a string:

```python
# Converting case
upper_case = my_string.upper()
lower_case = my_string.lower()
print(upper_case)
print(lower_case)
```

Here, we've converted the original string `"Hello, world!"` to uppercase using the `upper()` method, resulting in `"HELLO, WORLD!"`. We've also converted the string to lowercase using the `lower()` method, resulting in `"hello, world!"`. These operations are known as case conversion. 

Strings are incredibly versatile, offering a wide range of methods for manipulating and extracting data. This flexibility makes them invaluable for working with textual data, such as user input, file contents, or any text-based information you encounter in your code.

# 5. Making decisions and repeating things

## 5.1 Making decisions with `if`

An `if` statement runs some code only when a condition is `True`. The condition is usually a comparison (Section 4.1).

```python
turbidity = 35.0  # NTU

if turbidity > 50:
    print("High")
elif turbidity > 10:
    print("Moderate")
else:
    print("Low")
```

A few rules to notice:

- The line with `if` ends in a colon `:`.
- The code that belongs to the `if` is **indented** (pushed in by four spaces). Colab does this for you when you press Enter after a colon. Indentation is how Python knows which lines are inside the `if`.
- `elif` (short for "else if") checks another condition if the first was `False`. You can have as many as you need.
- `else` catches everything left over.

Python checks the conditions from top to bottom and runs only the **first** one that's `True`. Change `turbidity` to a few different values and run the cell again.

## 5.2 Repeating things with `for`

A `for` loop runs the same code once for each item in a list.

```python
readings = [8.2, 12.5, 6.9, 40.1]

for value in readings:
    print(value * 2)
```

Each time round the loop, `value` holds the next item from `readings`. The indented line runs once per item. You can call the loop variable anything you like, but a name that says what it holds makes your code easier to read.

You can loop over a dictionary too. `.items()` gives you each key and its value together.

```python
site_counts = {"Upstream": 8, "Downstream": 9}

for site, count in site_counts.items():
    print(f"{site} has {count} samples")
```

## 5.3 A shortcut: list comprehensions

Making a new list from an old one is so common that Python has a one-line shortcut for it, called a **list comprehension**. These two cells do the same thing.

```python
# With a for loop
doubled = []
for value in readings:
    doubled.append(value * 2)
print(doubled)
```

```python
# With a list comprehension
doubled = [value * 2 for value in readings]
print(doubled)
```

Read the second one as "`value * 2` for each `value` in `readings`". You don't have to use list comprehensions, but you'll see them in other people's code (and in code from GenAI), so it's worth recognising them.

# 6. Functions, methods and arguments

You've already used quite a few **functions**: `print()`, `type()`, `len()`, `sum()` and `range()`. A function is a named bit of code that does a job. You **call** it by writing its name followed by round brackets, and you put what it needs inside the brackets. The things you put inside are called **arguments**.

```python
# round() takes a number and how many decimal places to keep
print(round(3.14159, 2))
```

You'll see keyword arguments everywhere in data analysis. For example, `np.std(data, ddof=1)` in Practical 3, or `pd.read_csv(url, dayfirst=True)` in Practical 2.

A **method** is a function that belongs to a value. You call it with a dot after the value. You've already seen string methods like `.upper()` and `.replace()` in Section 4.5.

```python
station = "laidley creek"
print(station.title())
print(station.replace("creek", "Ck"))
```

# 8. A first look at NumPy

Lists are great for storing values, but they can't do maths on every value at once. Try subtracting 1 from a list of numbers:

```python
rainfall_mm = [12.4, 0.0, 3.2, 25.8, 7.1]

# Remove the # on the next line and run it. You'll get a TypeError.
# print(rainfall_mm - 1)
```

Python doesn't know how to take 1 away from a list. This is where **NumPy** (short for Numerical Python) comes in. It's a library: a bundle of extra tools you load with an `import` statement. By convention we load it with the short name `np`.

```python
import numpy as np
```

There are many functions in NumPy, a few examples are below:

- `np.array()`: Converts a list into a NumPy array.
- `np.sum()`: Calculates the sum of all elements in an array.
- `np.mean()`: Calculates the mean of all elements in an array.
- `np.sqrt()`: Calculates the square root of all elements in an array.
- `np.log()`: Calculates the natural logarithm of all elements in an array.
- `np.exp()`: Calculates the exponential of all elements in an array.
- `np.sin()`, `np.cos()`, `np.tan()`: Calculates the sine, cosine, and tangent of all elements in an array.
- `np.pi`: A constant representing the value of pi (3.14159...).
- `np.e`: A constant representing the value of Euler's number (2.71828...).
- `np.linspace()`: Generates a sequence of evenly spaced numbers over a specified range.
- `np.arange()`: Generates a sequence of numbers over a specified range with a specified step size.

Let's turn our rainfall list into a NumPy array and try again.

```python
rainfall_array = np.array(rainfall_mm)
print(rainfall_array - 1)
```

This time it works: NumPy took 1 off every value at once. You can do the same with any maths, which makes NumPy arrays really useful for statistics.

```python
# Every value squared
print(rainfall_array ** 2)

# Some handy summaries
print("Total:", np.sum(rainfall_array))
print("Mean:", np.mean(rainfall_array))
print("Count:", len(rainfall_array))
```

Comparisons work on whole arrays too. You get back an array of `True` and `False`, one for each value. Python counts `True` as 1 and `False` as 0, so summing it tells you how many values passed the test.

```python
print(rainfall_array > 5)
print("Days with more than 5 mm:", np.sum(rainfall_array > 5))
```
